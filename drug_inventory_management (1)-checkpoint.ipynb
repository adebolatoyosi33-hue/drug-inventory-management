# Drug Inventory Management System

### Python-Based Inventory Analysis and Decision-Support Prototype

This project develops a Python-based Drug Inventory Management System
for analysing simulated medicine inventory data.

The system demonstrates how Python can be used to:

- structure inventory records;
- validate data;
- calculate inventory indicators;
- identify stock-level risks;
- monitor expiry risks;
- perform inventory searches;
- generate management summaries; and
- translate analytical findings into practical stakeholder actions.

**Project type:** Educational / portfolio project  
**Data source:** Simulated inventory dataset  
**Core tools:** Python

## 1. Project Overview

Medicine inventory management requires timely information about stock
availability, inventory value, expiry risk and potential overstocking.

This project demonstrates how a simple Python-based analytical system can
transform structured inventory records into information that can support
inventory-management decisions.

The project follows the analytical workflow:

**Data → Validation → Processing → Analysis → Risk Identification → 
Management Summary → Stakeholder Action**

The system is an educational prototype and should not be considered a
replacement for professional pharmaceutical inventory-management systems.

## 2. Business Questions

The analysis is designed to answer the following practical inventory
questions:

1. How many medicines are being tracked?
2. What is the inventory value?
3. Which medicines are low stock?
4. Which medicines are critically low?
5. Which medicines are overstocked?
6. Which medicine has the highest inventory value?
7. Which medicine has the lowest stock?
8. How is inventory distributed across facilities?
9. Which medicines are approaching expiry?
10. Which inventory issues require management attention?
11. What actions should stakeholders consider based on the findings?

## 4. Data Validation

Data validation is performed before analysis to reduce the risk of
producing misleading results from incorrect inventory records.

The validation rules check for:

- negative quantities;
- negative unit prices;
- negative reorder levels;
- maximum stock below reorder level; and
- missing medicine names.

Records containing validation errors are excluded from the validated
inventory used for subsequent analysis.

```python
# --- A negative quantity cannot represent physical inventory.

for drug in inventory:
    errors = []

    if drug["quantity"] < 0:
        errors.append("Negative quantity")

    if drug["unit_price"] < 0:
        errors.append("Negative unit price")

    if drug["drug_name"] == "":
        errors.append("Missing drug name")

    if drug["maximum_stock"] < drug["reorder_level"]:
        errors.append("Maximum stock below reorder level")

    if errors:
        print(drug["drug_id"], ":", errors)
```

### 4.1 Validation Interpretation

The validation process demonstrates why inventory data should not be
analysed blindly.

The intentionally invalid records allow the system to demonstrate how
common data-quality problems can be detected before they affect inventory
calculations and risk classifications.

```python
valid_inventory = []
invalid_inventory = []

for drug in inventory:
    errors = []

    if drug["quantity"] < 0:
        errors.append("Negative quantity")

    if drug["unit_price"] < 0:
        errors.append("Negative unit price")

    if drug["drug_name"] == "":
        errors.append("Missing drug name")

    if drug["maximum_stock"] < drug["reorder_level"]:
        errors.append("Maximum stock below reorder level")

    if errors:
        invalid_inventory.append(drug)
    else:
        valid_inventory.append(drug)

print("Total records:", len(inventory))
print("Valid records:", len(valid_inventory))
print("Invalid records:", len(invalid_inventory))

for drug in invalid_inventory:
    print(drug["drug_id"], "-", drug["drug_name"])
```

## 5. Inventory Value Analysis

Inventory value represents the estimated monetary value of stock held for
each medicine.

**Formula:**

Inventory Value = Quantity × Unit Price

The calculation is applied to each valid inventory record and is then used
to derive overall inventory-value indicators.

```python
# ============================================================
# INVENTORY VALUE CALCULATION
# ============================================================

def calculate_inventory_value(drug):
    """
    Calculate the total inventory value for one medicine.

    Parameters:
        drug (dict): A validated inventory record.

    Returns:
        float/int: Total value of the medicine in stock.
    """
    
    quantity = drug["quantity"]
    unit_price = drug["unit_price"]
    
    return quantity * unit_price

# Test the inventory value calculation using the first valid record.

first_drug = valid_inventory[0]

inventory_value = calculate_inventory_value(first_drug)

print("Drug:", first_drug["drug_name"])
print("Quantity:", first_drug["quantity"])
print("Unit Price: ₦", first_drug["unit_price"])
print("Inventory Value: ₦", inventory_value)
```

### Inventory Value for All Valid Medicines

The following analysis calculates the estimated inventory value
for each valid medicine record.

Invalid records are excluded because they contain data-quality
problems that could distort the calculations.

```python
# ============================================================
# CALCULATE INVENTORY VALUE FOR EACH VALID MEDICINE
# ============================================================

for drug in valid_inventory:
    
    # Calculate the value of the current stock.
    inventory_value = calculate_inventory_value(drug)
    
    # Display the medicine and its calculated inventory value.
    print(
        drug["drug_name"],
        "→ ₦",
        inventory_value
    )
```

## 5. Total Inventory Value

The total inventory value is calculated by summing the inventory
value of every valid medicine record.

Formula:

Total Inventory Value = Σ(Quantity × Unit Price)

```python
# ============================================================
# TOTAL INVENTORY VALUE
# ============================================================

# Start the accumulator at zero.
total_inventory_value = 0

# Add the inventory value of each valid medicine.
for drug in valid_inventory:
    
    inventory_value = calculate_inventory_value(drug)
    
    total_inventory_value += inventory_value

# Display the final result.
print("Total Inventory Value: ₦", total_inventory_value)
```

## 6. Stock Status Analysis

Stock levels are classified using the inventory thresholds defined for
this project.

| Status | Rule |
|---|---|
| Critical | Quantity ≤ 50% of Reorder Level |
| Low Stock | Quantity ≤ Reorder Level |
| Overstocked | Quantity > Maximum Stock |
| Normal | Does not meet the conditions above |

These classifications are simplified analytical rules used for this
educational prototype.

```python
# ============================================================
# STOCK STATUS CLASSIFICATION
# ============================================================

def classify_stock(drug):
    """
    Classify the stock status of a medicine.

    Parameters:
        drug (dict): A validated inventory record.

    Returns:
        str: Stock status of the medicine.
    """

    quantity = drug["quantity"]
    reorder_level = drug["reorder_level"]
    maximum_stock = drug["maximum_stock"]

    # Critical: stock is at or below 50% of the reorder level.
    if quantity <= 0.5 * reorder_level:
        return "Critical"

    # Low stock: quantity is below or equal to the reorder level.
    elif quantity <= reorder_level:
        return "Low Stock"

    # Overstock: quantity is above the maximum stock level.
    elif quantity > maximum_stock:
        return "Overstock"

    # Adequate: stock falls within the acceptable range.
    else:
        return "Adequate"
```

```python
# Test the stock classification using the first valid medicine.

first_drug = valid_inventory[0]

stock_status = classify_stock(first_drug)

print("Drug:", first_drug["drug_name"])
print("Quantity:", first_drug["quantity"])
print("Reorder Level:", first_drug["reorder_level"])
print("Maximum Stock:", first_drug["maximum_stock"])
print("Stock Status:", stock_status)
```

### 7. Stock Status for All Valid Medicines

```python
# ============================================================
# CLASSIFY ALL VALID INVENTORY RECORDS
# ============================================================

for drug in valid_inventory:

    # Determine the stock status for the current medicine.
    status = classify_stock(drug)

    # Display the medicine and its stock status.
    print(f"{drug['drug_name']}: {status}")
```

## 8. Stock Status Summary

This section summarizes the number of valid medicines falling
within each stock-status category.

```python
# ============================================================
# STOCK STATUS SUMMARY
# ============================================================

# Create counters for each stock category.
stock_summary = {
    "Critical": 0,
    "Low Stock": 0,
    "Adequate": 0,
    "Overstock": 0
}

# Classify each valid medicine and update the appropriate counter.
for drug in valid_inventory:

    status = classify_stock(drug)

    stock_summary[status] += 1

# Display the summary.
for status, count in stock_summary.items():
    print(f"{status}: {count}")
```

```python
# ============================================================
# STOCK STATUS CLASSIFICATION
# ============================================================

def classify_stock(drug):
    """
    Classify a medicine based on its current stock level.

    Returns:
        Critical, Low Stock, Adequate, or Overstock
    """

    quantity = drug["quantity"]
    reorder_level = drug["reorder_level"]
    maximum_stock = drug["maximum_stock"]

    # Critical stock: quantity is 50% or less of reorder level
    if quantity <= 0.5 * reorder_level:
        return "Critical"

    # Low stock: quantity is at or below reorder level
    elif quantity <= reorder_level:
        return "Low Stock"

    # Overstock: quantity is above maximum stock
    elif quantity > maximum_stock:
        return "Overstock"

    # Otherwise, stock is adequate
    else:
        return "Adequate"
```

```python
# Test the stock classification using Paracetamol

drug = valid_inventory[0]

status = classify_stock(drug)

print("Drug:", drug["drug_name"])
print("Stock Status:", status)
```

## 9. Medicines Requiring Reorder

Medicines classified as Critical or Low Stock require attention
because their current quantity is at or below the defined
reorder threshold.

```python
# ============================================================
# IDENTIFY MEDICINES REQUIRING REORDER
# ============================================================

for drug in valid_inventory:

    # Determine the current stock status.
    status = classify_stock(drug)

    # Display medicines that require replenishment.
    if status == "Critical" or status == "Low Stock":
        print(drug["drug_id"], "-", drug["drug_name"], "-", status)
```

## 10. Reorder Quantity

Reorder quantity estimates the number of units required to restore
a medicine's stock to its defined maximum stock level.

Formula:

Reorder Quantity = Maximum Stock − Current Quantity

This calculation is applied to medicines classified as Critical
or Low Stock.

```python
# ============================================================
# CALCULATE REORDER QUANTITY
# ============================================================

for drug in valid_inventory:

    # Determine the current stock status.
    status = classify_stock(drug)

    # Only calculate reorder quantity for medicines requiring action.
    if status == "Critical" or status == "Low Stock":

        # Calculate the number of units required to reach maximum stock.
        reorder_quantity = drug["maximum_stock"] - drug["quantity"]

        # Display the replenishment information.
        print(
            drug["drug_id"],
            "-",
            drug["drug_name"],
            "-",
            status,
            "- Reorder:",
            reorder_quantity,
            "units"
        )
```

## 11. Low-Stock Count

The low-stock count represents the number of inventory records

Rule:

Quantity <= Reorder Level

```python
# ============================================================
# LOW-STOCK COUNT
# ============================================================

# Start the counter at zero.
low_stock_count = 0

# Check every valid inventory record.
for drug in valid_inventory:

    # A medicine is considered low stock when its
    # current quantity is at or below the reorder level.
    if drug["quantity"] <= drug["reorder_level"]:
        low_stock_count += 1

# Display the required KPI.
print("Low-Stock Count:", low_stock_count)
```

## 12. Highest-Value Medicine

This section identifies the medicine with the highest total
inventory value.

**Formula:**

Inventory Value = Quantity × Unit Price

The analysis is performed using the 25 validated inventory records.

```python
# ============================================================
# HIGHEST-VALUE MEDICINE
# ============================================================

# Find the valid medicine with the highest inventory value.
highest_value_drug = max(
    valid_inventory,
    key=calculate_inventory_value
)

# Calculate its inventory value.
highest_value = calculate_inventory_value(highest_value_drug)

# Display the result.
print("Highest-Value Medicine")
print("----------------------")
print("Drug ID:", highest_value_drug["drug_id"])
print("Drug Name:", highest_value_drug["drug_name"])
print("Quantity:", highest_value_drug["quantity"])
print("Unit Price: ₦", highest_value_drug["unit_price"])
print("Inventory Value: ₦", highest_value)
```

## 14. Facility Analysis

This section examines the distribution of valid medicine records
across healthcare facilities.

The analysis counts how many inventory records are associated
with each facility.

```python
# ============================================================
# FACILITY ANALYSIS
# ============================================================

# Create an empty dictionary to store the facility counts.
facility_counts = {}

# Go through every valid medicine record.
for drug in valid_inventory:

    # Get the name of the healthcare facility.
    facility = drug["facility"]

    # If the facility is already in the dictionary,
    # increase its count by 1.
    if facility in facility_counts:
        facility_counts[facility] += 1

    # If this is the first record for the facility,
    # start its count at 1.
    else:
        facility_counts[facility] = 1


# Display the number of inventory records for each facility.
print("Inventory Records by Facility")
print("-----------------------------")

for facility, count in facility_counts.items():
    print(facility, ":", count)
```

# 24. Reusable Program Functions

The following functions organise the inventory operations into reusable
components. These functions will later be connected to the program's
interactive menu.

The main operations are:

- View inventory
- Search for a medicine
- Calculate inventory value
- Identify low-stock medicines
- Identify critical-stock medicines
- Identify overstocked medicines
- Identify medicines approaching expiry

```python
# ============================================================
# REUSABLE INVENTORY FUNCTIONS
# ============================================================


def view_inventory(inventory):
    """
    Display all medicine records in the supplied inventory list.
    """

    print("\nINVENTORY")
    print("=" * 70)

    for drug in inventory:
        print(
            f"{drug['drug_id']} | "
            f"{drug['drug_name']} | "
            f"Quantity: {drug['quantity']} | "
            f"Facility: {drug['facility']}"
        )


def search_medicine(inventory, search_term):
    """
    Search for medicines using the drug ID or medicine name.

    The search is case-insensitive.
    """

    search_term = search_term.lower()

    results = []

    for drug in inventory:

        drug_id = drug["drug_id"].lower()
        drug_name = drug["drug_name"].lower()

        if search_term in drug_id or search_term in drug_name:
            results.append(drug)

    return results


def get_low_stock_medicines(inventory):
    """
    Return medicines whose quantity is at or below
    their reorder level.
    """

    return [
        drug for drug in inventory
        if drug["quantity"] <= drug["reorder_level"]
    ]


def get_critical_stock_medicines(inventory):
    """
    Return medicines whose quantity is at or below
    50% of their reorder level.
    """

    return [
        drug for drug in inventory
        if drug["quantity"] <= 0.5 * drug["reorder_level"]
    ]


def get_overstocked_medicines(inventory):
    """
    Return medicines whose quantity is above their
    maximum stock level.
    """

    return [
        drug for drug in inventory
        if drug["quantity"] > drug["maximum_stock"]
    ]


def get_near_expiry_medicines(inventory, analysis_date, days=30):
    """
    Return medicines that will expire within the specified
    number of days from the analysis date.
    """

    near_expiry = []

    for drug in inventory:

        days_to_expiry = (
            drug["expiry_date"] - analysis_date
        ).days

        if days_to_expiry <= days:
            near_expiry.append(drug)

    return near_expiry
```

## 27. Inventory Summary Function

The inventory summary provides a concise overview of key stock,
valuation, expiry, and facility-level indicators.

```python
# ============================================================
# INVENTORY SUMMARY FUNCTION
# ============================================================

def display_inventory_summary(inventory, analysis_date):
    """
    Display a summary of the validated drug inventory.
    """

    # Basic inventory information
    total_records = len(inventory)

    facilities = set(
        drug["facility"] for drug in inventory
    )

    total_facilities = len(facilities)

    # Stock status
    low_stock = get_low_stock_medicines(inventory)

    critical_stock = get_critical_stock_medicines(inventory)

    overstocked = get_overstocked_medicines(inventory)

    # Expiry status
    near_expiry = get_near_expiry_medicines(
        inventory,
        analysis_date,
        NEAR_EXPIRY_DAYS
    )

    # Inventory values
    inventory_values = [
        calculate_inventory_value(drug)
        for drug in inventory
    ]

    total_inventory_value = sum(inventory_values)

    average_inventory_value = (
        total_inventory_value / total_records
    )

    # Average unit price
    total_unit_price = sum(
        drug["unit_price"]
        for drug in inventory
    )

    average_unit_price = (
        total_unit_price / total_records
    )

    # Highest-value medicine
    highest_value_drug = max(
        inventory,
        key=calculate_inventory_value
    )

    highest_value = calculate_inventory_value(
        highest_value_drug
    )

    # Display the summary
    print("\n")
    print("=" * 55)
    print("              INVENTORY SUMMARY")
    print("=" * 55)

    print(f"Total Inventory Records: {total_records}")
    print(f"Total Facilities: {total_facilities}")

    print("\nSTOCK STATUS")
    print("-" * 55)

    print(f"Low-Stock Medicines: {len(low_stock)}")
    print(f"Critical-Stock Medicines: {len(critical_stock)}")
    print(f"Overstocked Medicines: {len(overstocked)}")

    print("\nEXPIRY STATUS")
    print("-" * 55)

    print(f"Near-Expiry Medicines: {len(near_expiry)}")

    print("\nFINANCIAL SUMMARY")
    print("-" * 55)

    print(f"Average Unit Price: ₦{average_unit_price:,.2f}")

    print(
        f"Average Inventory Value: "
        f"₦{average_inventory_value:,.2f}"
    )

    print("\nHIGHEST-VALUE MEDICINE")
    print("-" * 55)

    print(
        f"Drug ID: {highest_value_drug['drug_id']}"
    )

    print(
        f"Drug Name: {highest_value_drug['drug_name']}"
    )

    print(
        f"Inventory Value: ₦{highest_value:,.2f}"
    )

    print("=" * 55)
```

```python
# ============================================================
# TEST INVENTORY SUMMARY
# ============================================================

display_inventory_summary(
    valid_inventory,
    analysis_date
)
```
